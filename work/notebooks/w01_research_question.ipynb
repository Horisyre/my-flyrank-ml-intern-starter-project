```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
else:
    # find the repo root from wherever this kernel started
    while not os.path.isdir("data/raw") and os.getcwd() != "/":
        os.chdir("..")

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
def correlation_coefficients(df,cols,target_col,method="pearson")-> pd.Series:
    corr = df[cols].corr(method=method)
    return corr[target_col][corr[target_col] > 0.7]

numeric_df = df.select_dtypes(include="number")
clicks_corr_pearson = correlation_coefficients(numeric_df,numeric_df.columns,"clicks_90d",method="pearson")

print(clicks_corr_pearson)
print("======================================================================================================================================")
clicks_corr_spearman = correlation_coefficients(numeric_df,numeric_df.columns,"clicks_90d",method="spearman")
print(clicks_corr_spearman)
print("======================================================================================================================================")


impressions_corr_pearson = correlation_coefficients(numeric_df,numeric_df.columns,"impressions_90d",method="pearson")
print(impressions_corr_pearson)
print("======================================================================================================================================") 
impressions_corr_spearman = correlation_coefficients(numeric_df,numeric_df.columns,"impressions_90d",method="spearman")
print(impressions_corr_spearman)
print("======================================================================================================================================")

engagement_corr_pearson = correlation_coefficients(numeric_df,numeric_df.columns,"engaged_sessions_90d",method="pearson")
print(engagement_corr_pearson)
print("======================================================================================================================================")
engagement_corr_spearman = correlation_coefficients(numeric_df,numeric_df.columns,"engaged_sessions_90d",method="spearman")
print(engagement_corr_spearman)
```
