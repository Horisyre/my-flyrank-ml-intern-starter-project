## 4. Data limits

### Unbalanced history

The dataset is not evenly distributed across months, with some months containing substantially more observations than others. This may affect model generalization if certain periods are overrepresented during training.

### Missing values

The dataset is checked for missing values to determine whether missing data could limit the use of the available variables for modelling.

### Uneven GSC and GA4 availability

GSC and GA4 availability is not necessarily consistent across all content records. Differences in availability may limit comparability between observations when Analytics-based variables are considered.

### Lack of daily observations

The dataset contains performance metrics aggregated over 30-day periods rather than daily performance observations. Therefore, it cannot capture day-to-day variation or identify precisely when performance changes occurred within a measurement window. This limits the ability to determine why performance changed based on daily or shorter-term behaviour.

### Available data fields

The available columns are inspected to identify the level of detail and types of variables present in the dataset. This helps determine what information is available for modelling and what information is absent from the dataset.
